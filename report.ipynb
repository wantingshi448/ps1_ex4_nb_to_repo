## Executive Summary

This report presents a focused analysis of the Titanic dataset to justify feature selection decisions for survival prediction. Using a Random Forest classifier, we identified **gender (Sex)** and **passenger class (Pclass)** as the two most critical features, accounting for 73.44% and 26.56% of predictive importance respectively. The following visualizations demonstrate why these features were prioritized and how they interact to influence survival outcomes.

---

## 1. Feature Importance: Data-Driven Selection

![Feature Importance](figures/1_feature_importance.png)

**Analysis:** The Random Forest model quantitatively confirms that **gender is the dominant predictor** (73.44% importance), with passenger class contributing the remaining 26.56%. This stark imbalance justifies focusing our analysis primarily on these two features rather than including less influential variables like age, fare, or embarkation port. The model's clear hierarchy guides our feature selection strategy: prioritize Sex and Pclass, as they capture the vast majority of predictable variance in survival outcomes.

**Feature Selection Decision:** Include Sex and Pclass as primary features; treat other variables as secondary or derive engineered features only if they show strong interaction effects with these core predictors.

---

## 2. Gender: The Most Critical Survival Factor

![Gender vs Survival](figures/2_gender_survival.png)

**Analysis:** The "women and children first" maritime protocol is starkly evident in the data. **Females had a 74.2% survival rate compared to just 18.9% for males**—a 55.3 percentage point difference that represents a 4x survival advantage. This massive disparity explains why gender alone accounts for nearly three-quarters of the model's predictive power. The absolute numbers reinforce this pattern: 233 of 314 females survived, while only 109 of 577 males did.

**Feature Selection Decision:** Gender is non-negotiable as a primary feature. Any model excluding this variable would sacrifice the majority of its predictive capability.

---

## 3. Passenger Class: Socioeconomic Stratification

![Passenger Class vs Survival](figures/3_pclass_survival.png)

**Analysis:** Passenger class reveals a clear **socioeconomic gradient in survival**: 1st class passengers had a 63.0% survival rate, declining to 47.3% for 2nd class and 24.2% for 3rd class. This 38.8 percentage point gap between highest and lowest classes reflects both physical location on the ship (upper decks had easier lifeboat access) and social dynamics during evacuation. The near-linear relationship between class and survival makes Pclass a powerful ordinal predictor.

**Feature Selection Decision:** Include Pclass as the secondary feature. Its moderate importance (26.56%) and clear relationship with survival complement gender's dominance while adding substantial additional predictive value.

---

## 4. Interaction Effects: Gender × Class Amplification

![Gender-Class Interaction](figures/4_gender_class_interaction.png)

**Analysis:** The interaction between gender and class reveals extreme outcome variation. **1st class females achieved an exceptional 96.8% survival rate**, while 3rd class males had only 13.5%—a staggering 83.3 percentage point difference. This demonstrates that these features don't operate independently; rather, they combine multiplicatively. Being both female and upper-class provided near-certain survival, while being male and lower-class meant near-certain death.

**Feature Selection Decision:** Model should capture the Sex × Pclass interaction term, as the combined effect exceeds the sum of individual effects. This justifies using both features together rather than selecting only one.

---

## 5. Age Distribution: Supporting Evidence for Core Features

![Age Analysis](figures/5_age_analysis.png)

**Analysis:** While age shows some predictive value—children (0-12) had a 58.0% survival rate versus 38.3% for young adults (19-35)—its effect is substantially weaker than gender and class. The age distributions of survivors and non-survivors show considerable overlap, suggesting age alone is insufficient for accurate prediction. The "children first" protocol existed but was less consistently enforced than "women first."

**Feature Selection Decision:** Treat age as a tertiary feature or derive a simple binary feature (child vs. adult) if model complexity permits. Age does not justify inclusion at the expense of computational resources given its limited incremental predictive value beyond Sex and Pclass.

---

## Conclusion

This focused analysis demonstrates that **parsimonious feature selection—prioritizing Sex and Pclass—captures the essential predictive signal** in the Titanic dataset. The visualizations show:

1. **Quantitative justification**: Random Forest importance scores (73.44% + 26.56% = 100%)
2. **Qualitative validation**: Clear survival disparities across gender (55.3pp gap) and class (38.8pp gap)  
3. **Interaction evidence**: Multiplicative effects justify including both features together
4. **Prioritization rationale**: Other variables show weaker or overlapping relationships

The resulting model achieved **76.54% validation accuracy** using primarily these two features, confirming that focused feature selection based on these visualizations produces a robust, interpretable predictor without unnecessary complexity.
